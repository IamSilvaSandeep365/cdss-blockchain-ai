```python
import json
import pickle
import os

DATA_DIR = '../data/ddxplus'

with open(os.path.join(DATA_DIR, 'release_conditions.json'), encoding='utf-8') as f:
    conditions = json.load(f)

# Build map: what the model outputs (the English condition name / abbreviation)
#            → the full readable English name
disease_name_map = {}
for key, details in conditions.items():
    eng = details.get('cond-name-eng', key)      # e.g. "URTI"
    # The model was trained on PATHOLOGY which matched cond-name-eng,
    # so map that value to itself if there's nothing fuller.
    disease_name_map[eng] = eng

# Now add nicer full-name overrides for the abbreviations
overrides = {
    "URTI": "Upper Respiratory Tract Infection (URTI)",
    "GERD": "Gastroesophageal Reflux Disease (GERD)",
    "HIV (initial infection)": "HIV (Initial Infection)",
    "PSVT": "Paroxysmal Supraventricular Tachycardia (PSVT)",
    "SLE": "Systemic Lupus Erythematosus (SLE)",
    "Possible NSTEMI / STEMI": "Possible NSTEMI / STEMI (Heart Attack)",
    "Bronchospasm / acute asthma exacerbation": "Bronchospasm / Acute Asthma Exacerbation",
    "Acute COPD exacerbation / infection": "Acute COPD Exacerbation / Infection",
}
disease_name_map.update(overrides)

with open('../models/disease_name_map.pkl', 'wb') as f:
    pickle.dump(disease_name_map, f)

print(f"✅ Disease name map saved with {len(disease_name_map)} entries")
print("\nSample:")
for k in ["URTI", "GERD", "PSVT", "Pneumonia"]:
    print(f"   {k:10} → {disease_name_map.get(k, k)}")
```
