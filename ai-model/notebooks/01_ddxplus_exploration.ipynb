```python
# Import Libraries

import pandas as pd
import numpy as np
import json
import zipfile
import os

pd.set_option('display.max_columns', None)
print("✅ Libraries imported!")
```

```python
#Load the JSON Files

DATA_DIR = '../data/ddxplus'

with open(os.path.join(DATA_DIR, 'release_conditions.json'), 'r', encoding='utf-8') as f:
    conditions = json.load(f)

with open(os.path.join(DATA_DIR, 'release_evidences.json'), 'r', encoding='utf-8') as f:
    evidences = json.load(f)

print(f"✅ Diseases (conditions) : {len(conditions)}")
print(f"✅ Evidences (symptoms)  : {len(evidences)}")
```

```python
#Explore the Disease List

print("=== First 5 Diseases ===\n")
disease_names = list(conditions.keys())

for name in disease_names[:5]:
    print(f"Disease: {name}")
    print(f"  Details: {conditions[name]}")
    print()

print(f"\n=== All {len(disease_names)} Diseases ===")
for i, name in enumerate(disease_names, 1):
    print(f"{i:2}. {name}")
```

```python
# Explore the Evidence (Symptom) Structure

evidence_keys = list(evidences.keys())

print("=== Sample Evidence Structure ===\n")
for key in evidence_keys[:3]:
    print(f"Evidence code: {key}")
    ev = evidences[key]
    print(f"  Question (EN) : {ev.get('question_en', 'N/A')}")
    print(f"  Data type     : {ev.get('data_type', 'N/A')}")
    print(f"  Possible values: {ev.get('possible-values', 'N/A')}")
    print()

print(f"Total evidence types: {len(evidence_keys)}")
```

```python
# Peek at the Training Patients

train_zip = os.path.join(DATA_DIR, 'release_train_patients.zip')

with zipfile.ZipFile(train_zip, 'r') as z:
    
    csv_name = z.namelist()[0]
    print(f"CSV inside zip: {csv_name}\n")
    
    with z.open(csv_name) as f:
        df_preview = pd.read_csv(f, nrows=1000)

print(f"Preview shape: {df_preview.shape}")
print(f"\nColumns: {df_preview.columns.tolist()}")
print(f"\n=== First 3 rows ===")
df_preview.head(3)
```
